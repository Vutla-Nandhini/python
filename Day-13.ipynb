```python
# finding even and odd numbers from alist
```

```python
lst=list(map(int,input().split()))
even=[]
odd=[]
for item in lst:
    if(item%2==0):
        even.append(item)
    else:
        odd.append(item)
print(even,odd)        
```

# patterns

Square pattern

```python
n=int(input())
for i in range(n):
    for j in range(n):
        print("*",end=" ")
    print()    
```

```python
n=int(input())
for i in range(n):
    for j in range(n):
        print(i+1,end=" ")   
    print()
```

```python
n=int(input())
for i in range(n):
    for j in range(n):
        print(j+1,end=" ")   
    print()
```

```python
n=int(input())
for i in range(n):
    for j in range(i,j+1):
        print("*",end=" ")   
    print()
```
