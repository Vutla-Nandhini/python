# set operations or set methods
1.union(new_set) : It returns the all unique elements from sets


```python
s1={1,2,3,6}
s2={3,4,5,6}
print(s1.union(s2))
print(s1|s2)
print("Intersection")
print(s1.intersection(s2))
print(s1 & s2)
```

```python
print("Difference")
print(s1.difference(s2))
print(s1-s2)
```

```python
print("symetric difference")
print(s1.symmetric_difference(s2))
```

```python
# add(ele) : It add to set, if the element is not present in set, otherwise nothing
# update(new_set): set elements updates with new set elements
# remove(ele) :It removes the set element, if element not present in set it raises keyError
# discard(ele): It removes the element form set otherwise nothing.
s={1,2,3,4}
s.add(10)
print(s)
s.update({1,2,5,6})
print(s)
s.remove(10)
print(s)
# s.remove(20) # keyError
print(s)
s.discard(10)
print(s)
print(s.discard(5))
print(s)
```
