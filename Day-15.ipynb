```python
# dictionary is ordered collection
details={'name':"nandhu",'Batch':49,"course":"python",'name':"srinu",0:1}
print(details[0])
details['course']='java'
details['age']=20
print(details)
```

```python
# dictionary methods
# get(key,default_value): It returns the value if the key is present in the dictionary
#   otherwise it returns default value
details={'name':"nandhu",'Batch':49,"course":"python",'name':"srinu",0:1}
print(details.get('location','india'))
print(details.get('name','ramu'))
print(details.get('marks'))
print(details)
```

```python
# update(items): Dictionary update with new items
details={'name':"nandhu",'Batch':49,"course":"python",'name':"srinu",0:1}
new_dict={'course':'java','Batch':50,'location':'india'}
details.update(new_dict)
print(details)
```

```python
# pop(key):It returns the values and removes the item from dictionary if key is present, otherwise key error
# popitem():It returns remove or last element
details = {'name':'Lydia','batch':49,0:1,'course':'python','name':'Kiran'}
print(details.pop('name'))
print(details.popitem())
print(details)
```
