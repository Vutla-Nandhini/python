# string operation
- string concatenation (using +)

- string repetition (using *)

```python
# string concatenation
s1 ='code'
s2 = 'gnan'
print(s1+" "+s2)
print((s1+" ")*5)
```

```python
# string concatenation
s1 ='code'
s2 = 'gnan'
print(s1+" "+s2)
print((s1+" ")*5)
```

```python
n=10
present=0
absent =0
for i in range(1,n+1):
    val = int(input(f'enter Roll no {i} attendence:(0-absent,1-present)'))
    if(val==1):
        present+=1
    else:
        absent+=1
per = present/n*100
print("The presenties are",present)
print("The absenties are",absent)
print("The persentage of presenties",per)
```
