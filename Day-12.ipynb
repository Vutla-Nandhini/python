# List Methods
- adding elements to list
- append(ele):Add the element at end of the list
- extand(list):adds all elements in list at end
- insert(pos,ele):Insert the element at specified position

```python
l=[1,2,3,4,5]
print(l.append(10)) #it returns None
print(l)
l.extend([11,12,13,14])
print(l)
l.insert(3,100)
print(l)
l.insert(100,1000)
print(l)
l.insert(-1,78)
print(l)
```

# removing elements from list
- remove(val) : it removes the first occurence value from list,otherwise valueError
- pop(): It remove,return end value from list,if list is empty it raises the error
- pop(index):It removes the specified element based on the index
- clear(): it removes all elements from list

```python
l=[1,2,3,4,5,6,5,7]
l.remove(5)
# l.remove(20) it returns value error
print(l)
l.pop()
print(l)
l.pop(1)
print(l)
l.clear()
print(l)
#l.pop()
del l
print(l)
```

```python
n=int(input())
abscou=0
precou=0
i=1
while(i!=n+1):
    value=int(input("Enter the value for absent/present: "))
    if(value==0):
        abscou+=1
        i+=1
    elif(value==1):
       precou+=1
       i+=1
    else:
        print("Invalid value")
print("total present",precou)
print("total absent",abscou)
print("prsent percent",precou//n*100)
```
