# functions


- Function it is block of statements, and function definition starts with 'def' keyword. It is a reusable block.
- The function defination executes whenever the function call occurs

```python
def fun_name(x):
    if x%2 == 0:
        return 'Even'
    else:
        return 'Odd'
num = int(input())        
print(fun_name(num))   
```

```python
def fun_name(x):
    if x%2 == 0:
        return 'Even'
    else:
        print("Hello")
num = int(input())        
print(fun_name(num))   
```

```python
def sum_fun(x):
    total=0
    for item in x:
        total+=item
    return total
lst=[1,2,3,4,5]
print(sum_fun(lst))
```

```python
# sum of two numbers
def sum_two_num(a,b):
    add = a+b
    return add
num1=int(input())
num2=int(input())
print(sum_two_num(num1,num2))
```

```python
# sum of three numbers
def sum_two_num(a,b,c):
    add = a+b+c
    return add
num1=int(input())
num2=int(input())
num3=int(input())
print(sum_two_num(num1,num2,num3))
```
