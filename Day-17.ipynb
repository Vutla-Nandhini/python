```python
# with exception - which blocks are going to execute----except,finally
try:
    a=10
    b=0
    print(a/b)
except:
    print("something wrong in program")
else:
    print("program done")
finally:
    print("succesfully program executed")
```

```python
# with exception - which blocks are going to execute----except,finally
try:
    a=10
    b=0
    print(a/b)
except ZeroDivisionError:
    print("Zero Division not possible")
except:
    print("something wrong in program")
else:
    print("program done")
finally:
    print("succesfully program executed")
```

```python
# correspond except block will execute
try:
    a=10
    b=0
    a.append[10]
    print(a/b)
except ZeroDivisionError:
    print("Zero Division not possible")
except:
    print("something wrong in program")
else:
    print("program done")
finally:
    print("succesfully program executed")
```
