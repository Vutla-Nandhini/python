```python
def add(a,b):
    return a+b
def sub(a,b):
    return a-b
def mul(a,b):
    return a*b
def div(a,b):
    return a//b
num1,num2 =map(int,input().split())
print("Enter the number 1.add,2.sub,3.mul,4.div")
n=int(input())
if(n==1):
    print(add(num1,num2))
elif(n==2):
    print(sub(num1,num2))
elif(n==3):
    print(mul(num1,num2))
else:
    print(div(num1,num2))
```

```python
# 2.Keyword arguments
def sub_fun(x,y):
    return x-y
a,b=map(int,input().split())  
print(sub_fun(x=a,y=b))
print(sub_fun(y=b,x=a))
```

```python
# sum of three numbers
def sum_two_num(a,b,c):
    add = a+b+c
    return add
num1=int(input())
num2=int(input())
num3=int(input())
print(sum_two_num(num1,num2,num3))
```

# variable length keyword arguments

```python
#variable length keyword arguments
def add(x, y=0, *args, **kwargs):
    print(x,y)
    print(type(args),args)
    print(type(kwargs),kwargs)
print(add(10,20,30,40,50,1,2,3,4,5,6,a=10, b=20, c=30))    
print(add(10))
print(add(10,-20))#,30,40,50,1,2,3,4,5,6,a=10, b=20, c=30))
print(add(10,20,30,40,50))#,1,2,3,4,5,6,a=10, b=20, c=30))
```

# List
- List is a collection different data type elements and it is mutable.
- list is defined with '[]'
# characteristics:
- List is ordered collection
- list is mutable
- list elements can access by using index
- list are sliceable
- list are heterogeneous datatype
- list is unhashable

```python
l=[1,2,3]+list(3)
```

```python
l=[1,2,3]+list[3]
```

```python
l=[1,2,3]+[3]
print(l)
```

```python
l=[1,2,3]+[[4,5]]
print(l)
```

```python
def add(x,*args):
    print(type(args))
    print(args)
    print(x)
print(add(10,20,30,40,50))    
```

```python
def add(x,y=0,*args,**kwargs):
    print(type(args),args)
    print(type(kwargs),kwargs)
    print(args)
    print(x,y)
print(add(10,20,30,40,50,a=1,b=2,c=3,d=4))    
print(add(10,-20))
print(add(10,20,30,40,50))
```
